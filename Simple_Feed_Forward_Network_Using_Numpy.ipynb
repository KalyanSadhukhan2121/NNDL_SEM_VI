```python
import numpy as np
x= np.array([0.8, 0.2, 0.3])
w= np.array([[0.2, 0.3],[0.5, 0.2],[0.3, 0.6]])
b= np.array([0.02,0.01])

z= np.dot(x, w) + b
print("Output: ", z)

def sigmoid(x):
  return 1/ (1+np.exp(-x))

a=sigmoid(z)
print("Activated Output: ", a)
```

```python
w2= np.array([[0.6],[0.3]])
b2= np.array([0.03])
z2= np.dot(a,w2)+b
a2=sigmoid(z2)
print("Final Output: ", a2)
```

```python
def relu(x):
  return np.maximum(0, x)

a_relu = relu(z)
print("ReLU Activated Output: ", a_relu)
a2_relu = relu(z2)
print("ReLU Final Output: ", a2_relu)
```
